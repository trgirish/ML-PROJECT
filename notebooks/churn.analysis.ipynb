```python
# Feature Engineering
x=data.drop("Churn",axis=1)
y=data["Churn"]
print("x shape:",x.shape)
print("y shape:",y.shape)
```

```python
x.columns
```

```python
x=x.drop("customerID",axis=1)
```

```python
x.columns
```

```python
numerical_features=x.select_dtypes(
    include=["int64","float64"]
).columns
categorical_features = x.select_dtypes(
    include=["str"]
).columns
```

```python
print("Numerical Features:")
print(numerical_features)

print("\nCategorical Features:")
print(categorical_features)
```

```python
# preprocessing
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split
```

```python
# 1. Convert Churn
data["Churn"] = data["Churn"].map({
    "No": 0,
    "Yes": 1
})
```

```python
x_train,x_test,y_train,y_test=train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
print("X_train:", x_train.shape)
print("X_test:", x_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
preprocessor = ColumnTransformer([
    ("numeric", StandardScaler(), numerical_features),
    ("categorical", OneHotEncoder(), categorical_features)
])
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.pipeline import Pipeline
```

```python
from sklearn.metrics import(
    accuracy_score,
    recall_score,
    precision_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)
```

```python

logistic_model=Pipeline(
    steps=[
        ("preprocessor",preprocessor),
        ("classifier",LogisticRegression(max_iter=1000))
    ]
)
```

```python
# model training
logistic_model.fit(x_train,y_train)
```

```python
# model prediction
logistic_prediction=logistic_model.predict(x_test)
logistic_prediction[:21]
```

```python
logistic_probabilities=logistic_model.predict_proba(x_test)[:, 1]
print(logistic_probabilities[:6])
```

```python
# comparing actual value and predicted value

print("Accuracy:",
      accuracy_score(y_test, logistic_prediction))

print("Precision:",
      precision_score(y_test, logistic_prediction,pos_label=1))

print("Recall:",
      recall_score(y_test, logistic_prediction,pos_label=1))

print("F1 Score:",
      f1_score(y_test, logistic_prediction,pos_label=1))

print("ROC-AUC:",
      roc_auc_score(y_test, logistic_probabilities))
```

```python
print(classification_report(y_test,logistic_prediction))
```

```python
cm=confusion_matrix(y_test,logistic_prediction)
print(cm)
```

```python
plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title("Logistic Regression - Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

```python
decision_tree_pipeline=Pipeline(
    steps=[
        ("preprocessor",preprocessor),
        (
            "model",
            DecisionTreeClassifier(
                class_weight="balanced",
                random_state=42
            )
        )
    ]
)
```

```python
decision_tree_pipeline.fit(x_train,y_train)
```

```python
decision_tree_prediction=decision_tree_pipeline.predict(x_test)
print(decision_tree_prediction[:10])
```

```python
decision_tree_probabilities = (
    decision_tree_pipeline.predict_proba(x_test)[:, 1]
)
print(decision_tree_probabilities)
```

```python
print(
    "Accuracy:",
    accuracy_score(
        y_test,
        decision_tree_prediction
    )
)

print(
    "Precision:",
    precision_score(
        y_test,
        decision_tree_prediction
    )
)

print(
    "Recall:",
    recall_score(
        y_test,
        decision_tree_prediction
    )
)

print(
    "F1 Score:",
    f1_score(
        y_test,
        decision_tree_prediction
    )
)

print(
    "ROC-AUC:",
    roc_auc_score(
        y_test,
        decision_tree_probabilities
    )
)
```

```python
random_forest_pipeline=Pipeline(
    steps=[
        ("preprocessor",preprocessor),
        (
            "model",
            RandomForestClassifier(
                n_estimators=200,
                class_weight="balanced",
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)
```

```python
random_forest_pipeline.fit(x_train,y_train)
```

```python
random_forest_prediction=(
    random_forest_pipeline.predict(x_test)
)
random_forest_prediction[:15]
```

```python
random_forest_probabilities = (
    random_forest_pipeline.predict_proba(x_test)[:, 1]
)
print(random_forest_probabilities)
```

```python
print(
    "Accuracy:",
    accuracy_score(
        y_test,
        random_forest_prediction
    )
)

print(
    "Precision:",
    precision_score(
        y_test,
        random_forest_prediction
    )
)

print(
    "Recall:",
    recall_score(
        y_test,
        random_forest_prediction
    )
)

print(
    "F1 Score:",
    f1_score(
        y_test,
        random_forest_prediction
    )
)

print(
    "ROC-AUC:",
    roc_auc_score(
        y_test,
        random_forest_probabilities
    )
)
```

```python
gradient_boosting_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            GradientBoostingClassifier(
                random_state=42
            )
        )
    ]
)
```

```python
gradient_boosting_pipeline.fit(
    x_train,
    y_train
)
```

```python
gradient_boosting_predictions = (
    gradient_boosting_pipeline.predict(x_test)
)
gradient_boosting_predictions[:14]
```

```python
gradient_boosting_probabilities = (
    gradient_boosting_pipeline.predict_proba(x_test)[:, 1]
)
gradient_boosting_probabilities[:13]
```

```python
print(
    "Accuracy:",
    accuracy_score(
        y_test,
        gradient_boosting_predictions
    )
)

print(
    "Precision:",
    precision_score(
        y_test,
        gradient_boosting_predictions
    )
)

print(
    "Recall:",
    recall_score(
        y_test,
        gradient_boosting_predictions
    )
)

print(
    "F1 Score:",
    f1_score(
        y_test,
        gradient_boosting_predictions
    )
)

print(
    "ROC-AUC:",
    roc_auc_score(
        y_test,
        gradient_boosting_probabilities
    )
)
```

```python
def evaluate_model(model_name, y_true, predictions, probabilities):
    return {
        "Model": model_name,
        "Accuracy": accuracy_score(
            y_true,
            predictions
        ),
        "Precision": precision_score(
            y_true,
            predictions
        ),
        "Recall": recall_score(
            y_true,
            predictions
        ),
        "F1": f1_score(
            y_true,
            predictions
        ),
        "ROC-AUC": roc_auc_score(
            y_true,
            probabilities
        )
    }
```

```python
results = []

results.append(
    evaluate_model(
        "Logistic Regression",
        y_test,
        logistic_prediction,
        logistic_probabilities
    )
)

results.append(
    evaluate_model(
        "Decision Tree",
        y_test,
        decision_tree_prediction,
        decision_tree_probabilities
    )
)

results.append(
    evaluate_model(
        "Random Forest",
        y_test,
        random_forest_prediction,
        random_forest_probabilities
    )
)

results.append(
    evaluate_model(
        "Gradient Boosting",
        y_test,
        gradient_boosting_predictions,
        gradient_boosting_probabilities
    )
)
```

```python
results_df = pd.DataFrame(results)
```
